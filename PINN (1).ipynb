```python
import airfrans as af
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import torch.nn.functional as F

import numpy as np
import pandas as pd
import copy
```

```python
train_data, training = af.dataset.load(root = 'DATA/Dataset', task = 'scarce', train = True)
test_data, testing = af.dataset.load(root = 'DATA/Dataset', task = 'scarce', train = False)
```

```python
class PINN(nn.Module):
    def __init__(self, input_dim = 2, hidden_dim = 128, hidden_layers = 5, output_dim = 3):
        super().__init__()
        layers = []

        layers.append(nn.Linear(input_dim, hidden_dim))
        layers.append(nn.Tanh())

        for _ in range(hidden_layers - 1):
            layers.append(nn.Linear(hidden_dim, hidden_dim))
            layers.append(nn.Tanh())

        layers.append(nn.Linear(hidden_dim, output_dim))
        self.network = nn.Sequential(*layers)


    # This function is to go across each network and tie the nodes together.
    def forward(self, coords):
        return self.network(coords)
```

```python
def physics_loss(pred, coords, targets_mean, targets_std, u_ref, l_ref):

    pred_physical = pred * targets_std + targets_mean 
    
    # Predicted values, the ones that we get from the model
    pred_u = pred_physical[:, 0:1]
    pred_v = pred_physical[:, 1:2]
    pred_p = pred_physical[:, 2:3]

    u_star = pred_u / u_ref
    v_star = pred_v / u_ref
    p_star = pred_p / (u_ref ** 2)

    # Get the partial derivatives for each of the values with respect to x and y
    grad_u = torch.autograd.grad(outputs = u_star, inputs = coords, grad_outputs = torch.ones_like(u_star), create_graph = True)[0]
    u_x = l_ref * grad_u[:, 0:1]
    u_y = l_ref * grad_u[:, 1:2]

    grad_v = torch.autograd.grad(outputs = v_star, inputs = coords, grad_outputs = torch.ones_like(v_star), create_graph = True)[0]
    v_x = l_ref * grad_v[:, 0:1]
    v_y = l_ref * grad_v[:, 1:2]

    grad_p = torch.autograd.grad(outputs = p_star, inputs = coords, grad_outputs = torch.ones_like(p_star), create_graph = True)[0]
    p_x = l_ref * grad_p[:, 0:1]
    p_y = l_ref * grad_p[:, 1:2]

    # Loss = 1/N SUMMATION(Rc^2 + Rx^2 + Ry^2)

    # Continuity equation: du/dx + du/dy = 0
    continuity = u_x + v_y

    # Momentum conservation equations: 
    # X-MOMENTUM = (u * du/dx) + (v * du/dy) * dP/dx
    # Y-MOMENTUM = (u * dv/dx) + (v * dv/dy) * dP/dy

    x_momentum = (u_star * u_x) + (v_star * u_y) + p_x
    y_momentum = (u_star * v_x) + (v_star * v_y) + p_y

    loss_continuity = torch.mean(continuity ** 2)
    loss_x_momentum = torch.mean(x_momentum ** 2)
    loss_y_momentum = torch.mean(y_momentum ** 2)

    loss = loss_continuity + loss_x_momentum + loss_y_momentum
    print(f"PHYS LOSS: {loss}")
    return loss
```

```python
simulation = train_data[2]
batch_size = 1024

coords = torch.tensor(simulation[:, 0:2], dtype = torch.float32)
targets = torch.tensor(simulation[:, 7:10], dtype = torch.float32)
```

```python
def data_loss(pred, targets):
    pred_u = pred[:, 0:1]
    pred_v = pred[:, 1:2]
    pred_p = pred[:, 2:3]

    real_u = targets[:, 0:1]
    real_v = targets[:, 1:2]
    real_p = targets[:, 2:3]

    loss_u = torch.mean((pred_u - real_u) ** 2)
    loss_v = torch.mean((pred_v - real_v) ** 2)
    loss_p = torch.mean((pred_p - real_p) ** 2)

    loss = loss_u + loss_v + loss_p
    print(f"DATA LOSS: {loss}")
    return loss
```

```python
model = PINN(input_dim = 2)
optimizer = optim.Adam(list(model.parameters()), lr = 0.001)

torch.manual_seed(42)
num_points = coords.shape[0]
all_indices = torch.randperm(num_points)
split = int(0.8 * num_points)

train_indices = all_indices[:split]
val_indices = all_indices[split:]

coords_train = coords[train_indices]
coords_val = coords[val_indices]

targets_train = targets[train_indices]
targets_val = targets[val_indices]


print("Training coordinates:", coords_train.shape)
print("Training targets:", targets_train.shape)

print("Validation coordinates:", coords_val.shape)
print("Validation targets:", targets_val.shape)

targets_mean = targets_train.mean(dim=0)
target_std = targets_train.std(dim=0)

targets_train_norm = (targets_train - targets_mean) / target_std

targets_val_norm = (targets_val - targets_mean) / target_std

u_inf_x = simulation[0, 2]
u_inf_y = simulation[0, 3]

u_ref = (u_inf_x ** 2 + u_inf_y ** 2) ** 0.5
l_ref = 1.0

num_epochs = 1000
for epochs in range(num_epochs):
    indicies = torch.randperm(coords_train.shape[0])[:batch_size]

    coords_batch = coords_train[indicies]
    target_batch = targets_train_norm[indicies]

    optimizer.zero_grad()

    pred = model(coords_batch)

    loss_data = data_loss(pred, target_batch)
    total_loss = loss_data

    total_loss.backward()
    optimizer.step()

lambda_phys = 0.068
optimizer = optim.Adam(list(model.parameters()), lr = 0.0005)
finetune_epochs = 1000
physics_batch = 512

data_history = []
physics_history = []

best_model_state = copy.deepcopy(model.state_dict())
max_data_loss = 0.50
best_physics_loss = float("inf")

for epochs in range(finetune_epochs):
    indices = torch.randperm(coords_train.shape[0])[:physics_batch]
    coords_batch = (coords_train[indices].clone().detach().requires_grad_(True))
    target_batch = targets_train_norm[indices]

    optimizer.zero_grad()
    pred = model(coords_batch)

    loss_data = data_loss(pred, target_batch)
    loss_phys = physics_loss(pred, coords_batch, targets_mean, target_std, u_ref, l_ref)
    total_loss = loss_data + (lambda_phys * loss_phys)

    total_loss.backward()
    optimizer.step()

    data_history.append(loss_data.item())
    physics_history.append(loss_phys.item())

    if loss_data.item() <= max_data_loss:
        if loss_phys.item() < best_physics_loss:
            best_physics_loss = loss_phys.item()
            best_model_state = copy.deepcopy(model.state_dict())

            print(f"New best model | "f"Epoch {epochs + 1} | " f"Data: {loss_data.item():.4f} | "f"Physics: {loss_phys.item():.4f}")

    if (epochs + 1) % 100 == 0:
        print(f"Epoch {epochs + 1} | "f"Data: {loss_data.item():.4f} | "f"Physics: {loss_phys.item():.4f} | "f"Weighted Physics: "f"{(lambda_phys * loss_phys).item():.4f}")
        
model.load_state_dict(best_model_state)
print("Best model restored")
print(f"Best physics loss: {best_physics_loss:.6f}")
```

```python
model.eval()
with torch.no_grad():
    pred_val_norm = model(coords_val)
    val_u = torch.mean((pred_val_norm[:, 0] - targets_val_norm[:, 0]) ** 2)
    val_v = torch.mean((pred_val_norm[:, 1] - targets_val_norm[:, 1]) ** 2)
    val_p = torch.mean((pred_val_norm[:, 2] - targets_val_norm[:, 2]) ** 2)

    val_total = val_u + val_v + val_p
    pred_physical = pred_val_norm * target_std + targets_mean

    loss_u = torch.mean((pred_val_norm[:, 0] - targets_val_norm[:, 0]) ** 2)
    loss_v = torch.mean((pred_val_norm[:, 1] - targets_val_norm[:, 1]) ** 2)
    loss_p = torch.mean((pred_val_norm[:, 2] - targets_val_norm[:, 2]) ** 2)

print(f"U validation loss: {loss_u.item():.6f}")
print(f"V validation loss: {loss_v.item():.6f}")
print(f"P validation loss: {loss_p.item():.6f}")
print(f"Total: {(loss_u + loss_v + loss_p).item():.6f}")


print("\nVALIDATION RESULTS")
print(f"U loss:     {val_u.item():.6f}")
print(f"V loss:     {val_v.item():.6f}")
print(f"P loss:     {val_p.item():.6f}")
print(f"Total loss: {val_total.item():.6f}")
```
